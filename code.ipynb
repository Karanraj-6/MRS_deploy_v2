```python
import ast
```

```python
def fetch_genres(text):
    l=[]
    for i in ast.literal_eval(text):
        l.append(i['name'])
    return l
```

```python
df['genres']=df['genres'].apply(fetch_genres)
```

```python
def fetch_production_comapanies(text):
    l=[]
    for i in ast.literal_eval(text):
        l.append(i['name'])
    return l
```

```python
df['production_companies']=df['production_companies'].apply(fetch_production_comapanies)
```

```python
def fetch_cast(text):
    l=[]
    counter=0
    for i in ast.literal_eval(text):
        if counter != 3:
            l.append(i['original_name'])
            counter+=1
        else:
            break
    return l
```

```python
df['cast']=df['cast'].apply(fetch_cast)
```

```python
df['crew'][7]
```

```python
import ast
import pandas as pd

def fetch_director(text):
    directors = []
    try:
        data_list = ast.literal_eval(text)
        
        if isinstance(data_list, list):
            for item in data_list:
                if item.get('department') == 'Directing':
                    directors.append(item.get('original_name'))
    except (ValueError, SyntaxError) as e:
        print(f"Error parsing input: {e}")
    return directors
```

```python
df['crew'] = df['crew'].apply(fetch_director)

# Delete rows where the directors list is empty

df = df[df['crew'].apply(lambda x: isinstance(x, list) and len(x) > 0)]
```

```python
df['original_language']=df['original_language'].apply(lambda x: [x])
```

```python
df
```

```python
import re

def remove_noise(text):
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    return text
```

```python

def remove_whitespaces(text):
    return [re.sub(r'\s+', '', word) for word in text]

df['cast']=df['cast'].apply(remove_whitespaces)
df['crew']=df['crew'].apply(remove_whitespaces)
```

```python
df['overview']=df['overview'].apply(remove_noise)
df['title_info']=df['title'].apply(remove_noise)
```

```python
df['overview']=df['overview'].apply(lambda X:X.split( ))
df['title_info']=df['title_info'].apply(lambda X:X.split( ))
```

```python
df['data']=df['title_info']+df['original_language']+df['cast']+df['crew']+df['production_companies']+df['genres']+df['overview']
```

```python
new_df=df[['movie_id','title','data']]
```

```python
new_df
```

```python
new_df=new_df.reset_index()
new_df=new_df.drop('index',axis=1)
```

```python
new_df
```

```python
new_df['data'].values
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

new_df['data'] = new_df['data'].apply(lambda text: [lemmatizer.lemmatize(x.lower()) for x in text])
```

```python
new_df
```

```python
new_df['data'] = new_df['data'].apply(lambda x: ' '.join(x))
```

```python
new_df['data'][1]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(stop_words='english',ngram_range=(1,5))

features_data=cv.fit_transform(new_df['data'])
```

```python
features_data.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
simularity=cosine_similarity(features_data)
simularity
```

```python
def recommendation(text):
    idx=new_df[new_df['title'] == text].index[0]
    cs=simularity[idx]
    neighbors=sorted(list(enumerate(cs)),reverse=True,key=lambda x:x[1])[1:11]
    
    for i in neighbors:
        print(new_df.iloc[i[0]].title)
```

```python
recommendation('Man of Steel')
```
